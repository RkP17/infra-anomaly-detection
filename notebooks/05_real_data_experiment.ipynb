## Load the dataset

```python
train1 = pd.read_csv("../data/AIClusterKPI/train1.csv")
train1.head()
```

```python
train2 = pd.read_csv("../data/AIClusterKPI/train2.csv")
train2.head()
```

```python
test = pd.read_csv("../data/AIClusterKPI/test.csv")
test.head()
```

```python
groundtruth = pd.read_csv("../data/AIClusterKPI/groundtruth.csv")
groundtruth.head()
```

### Check for missing values

```python
datasets = {
    "train1": train1,
    "train2": train2,
    "test": test,
    "groundtruth": groundtruth,
}

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("Columns:", df.columns.tolist())
    print(df.head(3).to_string(index=False))
    print("\nMissing values:", df.isna().sum().sum())
```

### Types

Timestamp is a `str`, convert into dateTime data type

```python
for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("Types:\n", df.dtypes)
    print("Describe:\n", df.describe())
```

```python
train1["timestamp"] = pd.to_datetime(train1["timestamp"])
print(train1["timestamp"].dtypes)
print(train1["timestamp"].isna().sum())

train2["timestamp"] = pd.to_datetime(train2["timestamp"])
print(train2["timestamp"].dtypes)
print(train2["timestamp"].isna().sum())
```

```python
groundtruth["start_time"], groundtruth["end_time"] = pd.to_datetime(groundtruth["start_time"]), pd.to_datetime(groundtruth["end_time"])

print(groundtruth["start_time"].dtypes)
print(groundtruth["start_time"].isna().sum())

print(groundtruth["end_time"].dtypes)
print(groundtruth["end_time"].isna().sum())
```

### Timestamp integrity 

```python
is_chronological = train1["timestamp"].is_monotonic_increasing
print(is_chronological)

intervals = train1["timestamp"].diff()
print(intervals.value_counts())

duplicated = train1["timestamp"].duplicated()
print(duplicated.sum())
```

```python
is_chronological = train2["timestamp"].is_monotonic_increasing
print(is_chronological)

intervals = train2["timestamp"].diff()
print(intervals.value_counts())

duplicated = train2["timestamp"].duplicated()
print(duplicated.sum())
```

### Check for unusual or invalid values
The dataset documentation mentions that negative CPU readings can arise from counter-reset or differential artefacts. This is around 2.4% of the dataset has negative CPU values 

```python
has_negative = (train1.select_dtypes(include=['number','int','float']) < 0).sum()
print(has_negative)
```

```python
has_negative = (train2.select_dtypes(include=['number','int','float']) < 0).sum()
print(has_negative)
```
